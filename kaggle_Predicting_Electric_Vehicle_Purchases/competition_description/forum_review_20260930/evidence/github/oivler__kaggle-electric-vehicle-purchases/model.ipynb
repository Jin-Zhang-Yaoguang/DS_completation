## Feature Engineering

```python
CAT_COLS = [
    "Gender",
    "City_Type",
    "Current_Car_Type",
    "Home_Charging_Possible",
    "Subsidy_Available",
    "Range_Anxiety_Level",
    "Income_bin",
]

PAIR_COLS = [
    ("Subsidy_Available", "Range_Anxiety_Level"),
    ("Subsidy_Available", "Home_Charging_Possible"),
    ("Subsidy_Available", "Environmental_Concern_Level"),
    ("Range_Anxiety_Level", "Home_Charging_Possible"),
    ("City_Type", "Current_Car_Type"),
    ("Subsidy_Available", "Income_bin"),
    ("Environmental_Concern_Level", "Income_bin"),
    ("Range_Anxiety_Level", "Income_bin"),
]

TRIPLE_COLS = [
    ("Subsidy_Available", "Environmental_Concern_Level", "Range_Anxiety_Level"),
    ("Subsidy_Available", "Environmental_Concern_Level", "Home_Charging_Possible"),
    ("Subsidy_Available", "Environmental_Concern_Level", "Income_bin"),
    ("Subsidy_Available", "Range_Anxiety_Level", "Income_bin"),
]

QUAD_COLS = [
    (
        "Subsidy_Available",
        "Environmental_Concern_Level",
        "Range_Anxiety_Level",
        "Income_bin",
    ),
    (
        "Subsidy_Available",
        "Environmental_Concern_Level",
        "Home_Charging_Possible",
        "Income_bin",
    ),
]

RANGE_ANXIETY_MAP = {"Low": 0, "Medium": 1, "High": 2}
INCOME_BINS = pd.qcut(
    train["Annual_Income_USD"],
    q=10,
    retbins=True,
    duplicates="drop",
)[1]
INC_DIVS = [100, 250, 500, 1000, 2000, 5000]
KM_DIVS = [5, 10, 25, 50]


def _trailing_zeros(a):
    z = np.zeros(len(a), dtype=np.int8)
    for k, m in enumerate([10, 100, 1000, 10000], start=1):
        z[a % m == 0] = k
    return z


def build_features(df):
    out = df.copy()
    out["Total_Charging"] = (
        out["Charging_Stations_Near_Home"] + out["Charging_Stations_Near_Work"]
    )
    out["Home_Work_Charging_Diff"] = (
        out["Charging_Stations_Near_Home"] - out["Charging_Stations_Near_Work"]
    )
    out["Home_Work_Charging_Ratio"] = out["Charging_Stations_Near_Home"] / (
        out["Charging_Stations_Near_Work"] + 1
    )
    out["Income_per_Car"] = out["Annual_Income_USD"] / (out["Number_of_Cars_Owned"] + 1)
    out["Commute_per_Car"] = out["Daily_Commute_km"] / (out["Number_of_Cars_Owned"] + 1)
    out["Stations_per_Car"] = out["Total_Charging"] / (out["Number_of_Cars_Owned"] + 1)
    out["Income_per_Commute"] = out["Annual_Income_USD"] / (out["Daily_Commute_km"] + 1)
    out["Has_Subsidy"] = (out["Subsidy_Available"] == "Yes").astype(np.int8)
    out["Can_Charge"] = (out["Home_Charging_Possible"] == "Yes").astype(np.int8)
    out["Anx_ord"] = out["Range_Anxiety_Level"].map(RANGE_ANXIETY_MAP)
    out["Low_Anx"] = (out["Anx_ord"] == 0).astype(np.int8)
    out["High_Anx"] = (out["Anx_ord"] == 2).astype(np.int8)
    out["Env"] = out["Environmental_Concern_Level"]
    out["High_Env"] = (out["Env"] >= 4).astype(np.int8)
    out["Very_High_Env"] = (out["Env"] >= 5).astype(np.int8)
    out["Income_k"] = out["Annual_Income_USD"] / 1000
    out["Income_log"] = np.log1p(out["Annual_Income_USD"])
    out["Commute_log"] = np.log1p(out["Daily_Commute_km"])
    out["Env_sq"] = out["Env"] ** 2
    out["Env_cb"] = out["Env"] ** 3
    out["Age_sq"] = out["Age"] ** 2
    out["Ready_to_Buy"] = out["Has_Subsidy"] * out["Can_Charge"] * out["Low_Anx"]
    out["Subsidy_Env"] = out["Has_Subsidy"] * out["Env"]
    out["Subsidy_Env_sq"] = out["Has_Subsidy"] * out["Env_sq"]
    out["Subsidy_High_Env"] = out["Has_Subsidy"] * out["High_Env"]
    out["No_Subsidy_High_Env"] = (1 - out["Has_Subsidy"]) * out["High_Env"]
    out["Subsidy_Income"] = out["Has_Subsidy"] * out["Income_k"]
    out["Subsidy_Charging"] = out["Has_Subsidy"] * out["Total_Charging"]
    out["Subsidy_Low_Anx"] = out["Has_Subsidy"] * out["Low_Anx"]
    out["Charge_Env"] = out["Can_Charge"] * out["Env"]
    out["Charge_Subsidy"] = out["Can_Charge"] * out["Has_Subsidy"]
    out["Env_Low_Anx"] = out["Env"] * out["Low_Anx"]
    out["Env_High_Anx"] = out["Env"] * out["High_Anx"]
    out["Income_Env"] = out["Income_k"] * out["Env"]
    out["Commute_Env"] = out["Daily_Commute_km"] * out["Env"]
    out["Stations_Env"] = out["Total_Charging"] * out["Env"]
    out["Age_Env"] = out["Age"] * out["Env"]
    out["Subsidy_Score"] = out["Has_Subsidy"] * (
        out["Env"] + out["Can_Charge"] + out["Low_Anx"] + out["High_Env"]
    )
    out["Wealthy_Green"] = (out["Income_k"] > 90).astype(np.int8) * out["Env"]
    out["Car_Light"] = (out["Number_of_Cars_Owned"] <= 1).astype(np.int8)
    out["Income_bin"] = pd.cut(
        out["Annual_Income_USD"],
        bins=INCOME_BINS,
        labels=False,
        include_lowest=True,
    )
    out["Income_bin"] = out["Income_bin"].fillna(-1).astype(int).astype(str)

    # --- income / commute digit structure (synthesizer mass points) ---
    inc = out["Annual_Income_USD"].to_numpy(np.int64)
    km10 = np.round(out["Daily_Commute_km"].to_numpy(float) * 10).astype(np.int64)
    out["inc_last_digit"] = (inc % 10).astype(np.int8)
    out["inc_mod_100"] = (inc % 100).astype(np.int16)
    out["inc_mod_1000"] = (inc % 1000).astype(np.int16)
    out["inc_mod_10000"] = (inc % 10000).astype(np.int32)
    out["inc_tz"] = _trailing_zeros(inc)
    out["inc_at_floor"] = (inc == 30000).astype(np.int8)
    out["inc_round_100"] = (inc % 100 == 0).astype(np.int8)
    out["inc_round_1000"] = (inc % 1000 == 0).astype(np.int8)
    out["km_last_digit"] = (km10 % 10).astype(np.int8)
    out["km_mod_100"] = (km10 % 100).astype(np.int8)
    out["km_at_floor"] = (km10 == 50).astype(np.int8)
    out["km_round_1"] = (km10 % 10 == 0).astype(np.int8)

    # Synthesizer income bands (non-monotone buy-rate dips/spikes)
    out["inc_band_60_65k"] = ((inc >= 60000) & (inc <= 65000)).astype(np.int8)
    out["inc_band_110_115k"] = ((inc >= 110000) & (inc <= 115000)).astype(np.int8)
    _pow = np.maximum(np.floor(np.log10(np.maximum(inc, 1))).astype(int), 0)
    out["inc_lead_digit"] = ((inc // (10 ** _pow)) % 10).astype(np.int8)

    for dv in INC_DIVS:
        out[f"inc_div_{dv}"] = (inc // dv).astype(np.int32)
    for dv in KM_DIVS:
        out[f"km_div_{dv}"] = (km10 // dv).astype(np.int16)

    # keys for nested TE + CatBoost cats
    out["Income_str"] = out["Annual_Income_USD"].astype(np.int64).astype(str)
    out["Income_div50_str"] = (inc // 50).astype(str)
    out["Income_div100_str"] = (inc // 100).astype(str)
    out["Income_div250_str"] = (inc // 250).astype(str)
    out["Income_div500_str"] = (inc // 500).astype(str)
    out["Income_div1000_str"] = (inc // 1000).astype(str)
    out["Km_str"] = km10.astype(str)
    out["Km_div5_str"] = (km10 // 5).astype(str)
    out["Subsidy_Income100_str"] = (
        out["Subsidy_Available"].astype(str) + "|" + (inc // 100).astype(str)
    )
    out["Env_Income100_str"] = (
        out["Environmental_Concern_Level"].astype(int).astype(str) + "|" + (inc // 100).astype(str)
    )
    out["Subsidy_Env_Income100_str"] = (
        out["Subsidy_Available"].astype(str)
        + "|"
        + out["Environmental_Concern_Level"].astype(int).astype(str)
        + "|"
        + (inc // 100).astype(str)
    )
    # Low-card ints as cats for CatBoost (embedding-like lookups)
    out["Age_str"] = out["Age"].astype(int).astype(str)
    out["Cars_str"] = out["Number_of_Cars_Owned"].astype(int).astype(str)
    out["Env_str"] = out["Environmental_Concern_Level"].astype(int).astype(str)
    out["StHome_str"] = out["Charging_Stations_Near_Home"].astype(int).astype(str)
    out["StWork_str"] = out["Charging_Stations_Near_Work"].astype(int).astype(str)

    out["Ladder_Score"] = (
        3.0 * out["Has_Subsidy"]
        + 1.2 * out["Env"]
        + 1.5 * out["Low_Anx"]
        + 0.6 * out["Can_Charge"]
        + 0.015 * out["Income_k"]
        - 0.8 * out["High_Anx"]
        - 0.004 * out["Daily_Commute_km"]
    )
    out["Top_Profile"] = (
        out["Has_Subsidy"]
        * out["Very_High_Env"]
        * out["Low_Anx"]
        * (out["Income_k"] > 100).astype(np.int8)
    )
    out["Blocked_Profile"] = (
        (1 - out["Has_Subsidy"]) * (1 - out["High_Env"]) * (1 - out["Low_Anx"])
    )
    out["Subsidy_Env_Income"] = out["Has_Subsidy"] * out["Env"] * out["Income_k"]
    out["Subsidy_Env_Income_sq"] = out["Has_Subsidy"] * out["Env_sq"] * out["Income_log"]
    out["Subsidy_Low_Anx_Income"] = out["Has_Subsidy"] * out["Low_Anx"] * out["Income_k"]

    for left, right in PAIR_COLS:
        pair_name = f"{left}__{right}"
        out[pair_name] = out[left].astype(str) + "_" + out[right].astype(str)

    for left, middle, right in TRIPLE_COLS:
        triple_name = f"{left}__{middle}__{right}"
        out[triple_name] = (
            out[left].astype(str)
            + "_"
            + out[middle].astype(str)
            + "_"
            + out[right].astype(str)
        )

    for a, b, c, d in QUAD_COLS:
        quad_name = f"{a}__{b}__{c}__{d}"
        out[quad_name] = (
            out[a].astype(str)
            + "_"
            + out[b].astype(str)
            + "_"
            + out[c].astype(str)
            + "_"
            + out[d].astype(str)
        )

    feature_cols = [c for c in out.columns if c not in ["id", "Will_Buy_EV"]]
    return out[feature_cols]


X = build_features(train)
X_test = build_features(test)
y = (train["Will_Buy_EV"] == "Yes").astype(int).values
PAIR_FEATURE_COLS = [f"{left}__{right}" for left, right in PAIR_COLS]
TRIPLE_FEATURE_COLS = [f"{left}__{middle}__{right}" for left, middle, right in TRIPLE_COLS]
QUAD_FEATURE_COLS = [f"{a}__{b}__{c}__{d}" for a, b, c, d in QUAD_COLS]
INCOME_TE_COLS = [
    "Income_str",
    "Income_div50_str",
    "Income_div100_str",
    "Income_div250_str",
    "Income_div500_str",
    "Income_div1000_str",
    "Km_str",
    "Km_div5_str",
    "Subsidy_Income100_str",
    "Env_Income100_str",
    "Subsidy_Env_Income100_str",
]
# multi-prior nested TE on the strongest keys only (cost control)
INCOME_TE_MULTI_PRIOR = [
    "Income_str",
    "Income_div100_str",
    "Subsidy_Env_Income100_str",
]
INCOME_TE_PRIORS = (100.0, 200.0, 400.0)

ENCODE_COLS = CAT_COLS + PAIR_FEATURE_COLS + TRIPLE_FEATURE_COLS + QUAD_FEATURE_COLS
INT_AS_CAT_COLS = ["Age_str", "Cars_str", "Env_str", "StHome_str", "StWork_str"]
CATBOOST_CAT_COLS = ENCODE_COLS + INCOME_TE_COLS + INT_AS_CAT_COLS
RESIDUAL_GROUP_COLS = ["Subsidy_Available", "Environmental_Concern_Level"]

# Transductive income structure (train+test; no labels)
_pooled_inc = pd.concat(
    [train["Annual_Income_USD"], test["Annual_Income_USD"]]
).to_numpy(np.float64)
_sorted_inc = np.sort(_pooled_inc)
_vc_exact = pd.Series(_pooled_inc).value_counts()

def _map_freq(series, vc):
    return series.map(vc).astype(np.float32)

def _add_round_freq(frame, raw, r, name):
    rounded = np.round(raw / r) * r
    vc = pd.Series(np.round(_pooled_inc / r) * r).value_counts()
    frame[name] = pd.Series(rounded).map(vc).astype(np.float32)

def _local_density(values, window):
    lo = np.searchsorted(_sorted_inc, values - window, side="left")
    hi = np.searchsorted(_sorted_inc, values + window, side="right")
    return (hi - lo).astype(np.float32)

for frame in (X, X_test):
    raw = frame["Annual_Income_USD"].to_numpy(np.float64)
    frame["freq_Annual_Income_USD"] = _map_freq(frame["Annual_Income_USD"], _vc_exact)
    frame["log_freq_income"] = np.log1p(frame["freq_Annual_Income_USD"])
    _add_round_freq(frame, raw, 50, "freq_income_r50")
    _add_round_freq(frame, raw, 100, "freq_income_r100")
    frame["inc_density_50"] = _local_density(raw, 50.0)
    frame["inc_density_100"] = _local_density(raw, 100.0)
    frame["log_inc_density_100"] = np.log1p(frame["inc_density_100"])

# SmoothKeys: pooled quantile bins (train+test), used as nested-TE / CB cats
_pooled_km = pd.concat([train["Daily_Commute_km"], test["Daily_Commute_km"]]).to_numpy(float)


def _qbin(values, pooled, nq, name_prefix, frame_tr, frame_te):
    edges = np.unique(np.quantile(pooled, np.linspace(0.0, 1.0, nq + 1)))
    if len(edges) < 3:
        return
    lab = [f"{name_prefix}{i}" for i in range(len(edges) - 1)]
    frame_tr[f"{name_prefix}q{nq}"] = pd.cut(
        values["tr"], bins=edges, labels=lab, include_lowest=True
    ).astype(str)
    frame_te[f"{name_prefix}q{nq}"] = pd.cut(
        values["te"], bins=edges, labels=lab, include_lowest=True
    ).astype(str)


_qbin(
    {"tr": X["Annual_Income_USD"], "te": X_test["Annual_Income_USD"]},
    _pooled_inc,
    32,
    "inc_",
    X,
    X_test,
)
_qbin(
    {"tr": X["Annual_Income_USD"], "te": X_test["Annual_Income_USD"]},
    _pooled_inc,
    64,
    "inc_",
    X,
    X_test,
)
_qbin(
    {"tr": X["Annual_Income_USD"], "te": X_test["Annual_Income_USD"]},
    _pooled_inc,
    128,
    "inc_",
    X,
    X_test,
)
_qbin(
    {"tr": X["Daily_Commute_km"], "te": X_test["Daily_Commute_km"]},
    _pooled_km,
    16,
    "km_",
    X,
    X_test,
)
_qbin(
    {"tr": X["Daily_Commute_km"], "te": X_test["Daily_Commute_km"]},
    _pooled_km,
    32,
    "km_",
    X,
    X_test,
)

X["Subsidy_IncQ64_str"] = X["Subsidy_Available"].astype(str) + "|" + X["inc_q64"].astype(str)
X_test["Subsidy_IncQ64_str"] = (
    X_test["Subsidy_Available"].astype(str) + "|" + X_test["inc_q64"].astype(str)
)
X["Env_IncQ64_str"] = (
    X["Environmental_Concern_Level"].astype(int).astype(str) + "|" + X["inc_q64"].astype(str)
)
X_test["Env_IncQ64_str"] = (
    X_test["Environmental_Concern_Level"].astype(int).astype(str)
    + "|"
    + X_test["inc_q64"].astype(str)
)

SMOOTH_TE_COLS = [
    "inc_q32",
    "inc_q64",
    "inc_q128",
    "km_q16",
    "km_q32",
    "Subsidy_IncQ64_str",
    "Env_IncQ64_str",
]
AGE_TE_COLS = ["Age_str"]
INCOME_TE_COLS = INCOME_TE_COLS + SMOOTH_TE_COLS + AGE_TE_COLS
INCOME_TE_MULTI_PRIOR = list(
    dict.fromkeys(INCOME_TE_MULTI_PRIOR + ["inc_q64", "Subsidy_IncQ64_str", "Age_str"])
)
CATBOOST_CAT_COLS = ENCODE_COLS + INCOME_TE_COLS + INT_AS_CAT_COLS
# de-dupe while preserving order
CATBOOST_CAT_COLS = list(dict.fromkeys(CATBOOST_CAT_COLS))

print(f"train rows: {len(train):,}")
print(f"test rows: {len(test):,}")
print(f"raw features: {X.shape[1]}")
print(f"encode keys: {len(ENCODE_COLS)}")
print(f"income/smooth TE keys: {len(INCOME_TE_COLS)}")
print(f"target rate: {y.mean():.6f}")
```

```python
def add_target_encoding(
    X_train,
    y_train,
    X_valid,
    X_test,
    cat_cols,
    smoothings=(5, 20, 80),
):
    global_mean = y_train.mean()
    X_train = X_train.copy()
    X_valid = X_valid.copy()
    X_test = X_test.copy()

    for col in cat_cols:
        stats = (
            pd.DataFrame({"target": y_train, col: X_train[col]})
            .groupby(col)["target"]
            .agg(["mean", "count"])
        )
        for smoothing in smoothings:
            stats[f"te_{smoothing}"] = (
                stats["count"] * stats["mean"] + smoothing * global_mean
            ) / (stats["count"] + smoothing)
            mapping = stats[f"te_{smoothing}"]
            X_train[f"{col}_te_{smoothing}"] = X_train[col].map(mapping).fillna(global_mean)
            X_valid[f"{col}_te_{smoothing}"] = X_valid[col].map(mapping).fillna(global_mean)
            X_test[f"{col}_te_{smoothing}"] = X_test[col].map(mapping).fillna(global_mean)

        X_train[f"{col}_cnt"] = X_train[col].map(stats["count"]).fillna(0)
        X_valid[f"{col}_cnt"] = X_valid[col].map(stats["count"]).fillna(0)
        X_test[f"{col}_cnt"] = X_test[col].map(stats["count"]).fillna(0)

    drop_cols = [c for c in cat_cols if c in X_train.columns]
    return (
        X_train.drop(columns=drop_cols),
        X_valid.drop(columns=drop_cols),
        X_test.drop(columns=drop_cols),
    )


def add_nested_income_te(
    X_train,
    y_train,
    frames,
    cols,
    prior=200.0,
    n_splits=5,
    seed=7,
    suffix=None,
):
    """Cross-fitted TE for high-card income/commute keys (no row encodes itself)."""
    global_mean = float(np.mean(y_train))
    X_train = X_train.copy()
    outs = [f.copy() for f in frames]
    y_arr = np.asarray(y_train)
    skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=seed)
    inner = list(skf.split(np.zeros(len(y_arr)), y_arr))
    tag = "nte" if suffix is None else f"nte_{suffix}"

    for col in cols:
        enc = np.full(len(y_arr), global_mean, dtype=np.float64)
        keys = X_train[col].astype(str).to_numpy()
        for a, b in inner:
            s = pd.DataFrame({"k": keys[a], "y": y_arr[a]}).groupby("k")["y"].agg(["sum", "count"])
            sm = (s["sum"] + prior * global_mean) / (s["count"] + prior)
            enc[b] = pd.Series(keys[b]).map(sm).fillna(global_mean).to_numpy()
        X_train[f"{col}_{tag}"] = enc.astype(np.float32)
        s = pd.DataFrame({"k": keys, "y": y_arr}).groupby("k")["y"].agg(["sum", "count"])
        mapping = (s["sum"] + prior * global_mean) / (s["count"] + prior)
        for out in outs:
            out[f"{col}_{tag}"] = (
                out[col].astype(str).map(mapping).fillna(global_mean).astype(np.float32)
            )
    return (X_train, *outs)


def add_group_income_residuals(X_train, X_valid, X_test, group_cols):
    X_train = X_train.copy()
    X_valid = X_valid.copy()
    X_test = X_test.copy()

    group_key = X_train[group_cols[0]].astype(str)
    for col in group_cols[1:]:
        group_key = group_key + "|" + X_train[col].astype(str)
    train_tmp = X_train.copy()
    train_tmp["_group"] = group_key
    group_stats = train_tmp.groupby("_group")["Annual_Income_USD"].agg(["mean", "std", "median"])
    group_stats["std"] = group_stats["std"].fillna(group_stats["std"].median())

    def _apply(df):
        out = df.copy()
        key = out[group_cols[0]].astype(str)
        for col in group_cols[1:]:
            key = key + "|" + out[col].astype(str)
        out["_group"] = key
        out["Income_group_mean"] = out["_group"].map(group_stats["mean"])
        out["Income_group_median"] = out["_group"].map(group_stats["median"])
        out["Income_group_std"] = out["_group"].map(group_stats["std"]).replace(0, np.nan)
        global_mean = X_train["Annual_Income_USD"].mean()
        global_std = X_train["Annual_Income_USD"].std()
        out["Income_group_mean"] = out["Income_group_mean"].fillna(global_mean)
        out["Income_group_median"] = out["Income_group_median"].fillna(global_mean)
        out["Income_group_std"] = out["Income_group_std"].fillna(global_std)
        out["Income_resid_group"] = out["Annual_Income_USD"] - out["Income_group_mean"]
        out["Income_z_group"] = out["Income_resid_group"] / out["Income_group_std"]
        out["Income_above_group_med"] = (
            out["Annual_Income_USD"] > out["Income_group_median"]
        ).astype(np.int8)
        return out.drop(columns=["_group"])

    return _apply(X_train), _apply(X_valid), _apply(X_test)


def get_cat_idx(columns, cat_cols):
    return [columns.get_loc(col) for col in cat_cols if col in columns]



def add_cell_income_isotonic(
    X_train,
    y_train,
    frames,
    group_cols=("Subsidy_Available", "Environmental_Concern_Level"),
    n_splits=5,
    seed=17,
):
    """Leak-safe isotonic income -> buy rate inside each subsidy x env cell."""
    from sklearn.isotonic import IsotonicRegression

    X_train = X_train.copy()
    outs = [f.copy() for f in frames]
    y_arr = np.asarray(y_train)
    g_tr = X_train[list(group_cols)].astype(str).agg("|".join, axis=1).to_numpy()
    inc_tr = X_train["Annual_Income_USD"].to_numpy(float)
    prior = float(y_arr.mean())
    enc = np.full(len(y_arr), prior, dtype=np.float64)

    skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=seed)
    for a, b in skf.split(np.zeros(len(y_arr)), y_arr):
        models = {}
        for key in np.unique(g_tr[a]):
            m = g_tr[a] == key
            if m.sum() < 80 or len(np.unique(y_arr[a][m])) < 2:
                continue
            iso = IsotonicRegression(out_of_bounds="clip")
            iso.fit(inc_tr[a][m], y_arr[a][m])
            models[key] = iso
        for i, idx in enumerate(b):
            iso = models.get(g_tr[idx])
            enc[idx] = float(iso.predict([inc_tr[idx]])[0]) if iso is not None else prior

    X_train["cell_iso_income"] = enc.astype(np.float32)

    # full-train maps for valid/test
    models = {}
    for key in np.unique(g_tr):
        m = g_tr == key
        if m.sum() < 80 or len(np.unique(y_arr[m])) < 2:
            continue
        iso = IsotonicRegression(out_of_bounds="clip")
        iso.fit(inc_tr[m], y_arr[m])
        models[key] = iso

    for out in outs:
        g = out[list(group_cols)].astype(str).agg("|".join, axis=1).to_numpy()
        inc = out["Annual_Income_USD"].to_numpy(float)
        vals = np.full(len(out), prior, dtype=np.float64)
        for key, iso in models.items():
            m = g == key
            if m.any():
                vals[m] = iso.predict(inc[m])
        out["cell_iso_income"] = vals.astype(np.float32)

    return (X_train, *outs)


def prepare_te_frames(
    X_tr, y_tr, X_va, X_te, drop_income_keys=True, encode_cols=None
):
    encode_cols = ENCODE_COLS if encode_cols is None else encode_cols
    X_tr, X_va, X_te = add_group_income_residuals(
        X_tr, X_va, X_te, RESIDUAL_GROUP_COLS
    )
    X_tr, X_va, X_te = add_cell_income_isotonic(X_tr, y_tr, [X_va, X_te])
    # default prior=200 on all income TE keys
    X_tr, X_va, X_te = add_nested_income_te(
        X_tr, y_tr, [X_va, X_te], INCOME_TE_COLS, prior=200.0, suffix="200"
    )
    # extra priors on strongest keys
    for prior in INCOME_TE_PRIORS:
        if abs(prior - 200.0) < 1e-9:
            continue
        X_tr, X_va, X_te = add_nested_income_te(
            X_tr,
            y_tr,
            [X_va, X_te],
            INCOME_TE_MULTI_PRIOR,
            prior=prior,
            suffix=str(int(prior)),
        )
    X_tr, X_va, X_te = add_target_encoding(X_tr, y_tr, X_va, X_te, encode_cols)
    if drop_income_keys:
        drop = [c for c in INCOME_TE_COLS if c in X_tr.columns]
        X_tr = X_tr.drop(columns=drop, errors="ignore")
        X_va = X_va.drop(columns=drop, errors="ignore")
        X_te = X_te.drop(columns=drop, errors="ignore")
    return X_tr, X_va, X_te


def power_mean(arrays, p):
    stacked = np.stack(arrays, axis=0)
    stacked = np.clip(stacked, 1e-7, 1 - 1e-7)
    if abs(p) < 1e-9:
        return np.exp(np.mean(np.log(stacked), axis=0))
    return np.mean(stacked ** p, axis=0) ** (1.0 / p)


LGB_SPLITS = 10
LGB_SEEDS = [42, 2024, 7]
CB_SPLITS = 5
CB_SEEDS = [42, 2024]
LEAN_SEED = 7

LEAN_ENCODE_COLS = CAT_COLS
LEAN_CB_CATS = list(
    dict.fromkeys(CAT_COLS + INT_AS_CAT_COLS + ["Income_str", "Income_div100_str"])
)

LGB_PARAMS = {
    "n_estimators": 8000,
    "learning_rate": 0.02,
    "num_leaves": 63,
    "max_depth": 8,
    "min_child_samples": 24,
    "subsample": 0.65,
    "subsample_freq": 1,
    "colsample_bytree": 0.45,
    "reg_alpha": 0.14,
    "reg_lambda": 1.0,
    "max_bin": 8191,
    "path_smooth": 1.0,
}

CB_PARAMS = {
    "iterations": 4000,
    "learning_rate": 0.03,
    "depth": 8,
    "l2_leaf_reg": 4,
    "random_strength": 1,
    "bagging_temperature": 0.4,
    "border_count": 254,
    "max_ctr_complexity": 4,
    "loss_function": "Logloss",
    "eval_metric": "AUC",
    "verbose": False,
    "allow_writing_files": False,
}

LEAN_CB_PARAMS = {
    **CB_PARAMS,
    "learning_rate": 0.04,
    "depth": 7,
    "iterations": 3500,
    "random_seed": LEAN_SEED,
}

XGB_SPLITS = 5
XGB_SEEDS = [42, 7]
XGB_PARAMS = {
    "n_estimators": 6000,
    "learning_rate": 0.02,
    "max_depth": 7,
    "min_child_weight": 24,
    "subsample": 0.7,
    "colsample_bytree": 0.5,
    "reg_alpha": 0.12,
    "reg_lambda": 1.0,
    "tree_method": "hist",
    "eval_metric": "auc",
    "early_stopping_rounds": 200,
}

seed_oof = {seed: np.zeros(len(train)) for seed in LGB_SEEDS}
seed_test = {seed: np.zeros(len(test)) for seed in LGB_SEEDS}
cb_seed_oof = {seed: np.zeros(len(train)) for seed in CB_SEEDS}
cb_seed_test = {seed: np.zeros(len(test)) for seed in CB_SEEDS}
oof_lean = np.zeros(len(train))
test_lean = np.zeros(len(test))
xgb_seed_oof = {seed: np.zeros(len(train)) for seed in XGB_SEEDS}
xgb_seed_test = {seed: np.zeros(len(test)) for seed in XGB_SEEDS}
lgb_scores = []
cb_scores = []
lean_scores = []
xgb_scores = []

CELL_KEY_TRAIN = (
    train["Subsidy_Available"].astype(str)
    + "|"
    + train["Environmental_Concern_Level"].astype(int).astype(str)
).to_numpy()
CELL_KEY_TEST = (
    test["Subsidy_Available"].astype(str)
    + "|"
    + test["Environmental_Concern_Level"].astype(int).astype(str)
).to_numpy()


def _drop_strings_for_trees(*frames):
    out = []
    drop_base = list(dict.fromkeys(ENCODE_COLS + INT_AS_CAT_COLS + INCOME_TE_COLS))
    for f in frames:
        obj = f.select_dtypes(include=["object", "string", "category"]).columns.tolist()
        drop = [c for c in drop_base if c in f.columns] + obj
        out.append(f.drop(columns=list(dict.fromkeys(drop)), errors="ignore"))
    return out


print(f"training {len(LGB_SEEDS) * LGB_SPLITS} lightgbm models")
for seed in LGB_SEEDS:
    skf = StratifiedKFold(n_splits=LGB_SPLITS, shuffle=True, random_state=seed)
    for fold, (train_idx, valid_idx) in enumerate(skf.split(X, y), start=1):
        X_tr, X_va = X.iloc[train_idx], X.iloc[valid_idx]
        y_tr, y_va = y[train_idx], y[valid_idx]
        X_tr, X_va, X_te = prepare_te_frames(
            X_tr, y_tr, X_va, X_test, encode_cols=LEAN_ENCODE_COLS
        )
        X_tr, X_va, X_te = _drop_strings_for_trees(X_tr, X_va, X_te)

        model = lgb.LGBMClassifier(**LGB_PARAMS, random_state=seed, n_jobs=-1, verbose=-1)
        model.fit(
            X_tr,
            y_tr,
            eval_set=[(X_va, y_va)],
            eval_metric="auc",
            callbacks=[lgb.early_stopping(200, verbose=False), lgb.log_evaluation(0)],
        )
        valid_preds = model.predict_proba(X_va)[:, 1]
        fold_auc = roc_auc_score(y_va, valid_preds)
        seed_oof[seed][valid_idx] = valid_preds
        seed_test[seed] += model.predict_proba(X_te)[:, 1] / LGB_SPLITS
        lgb_scores.append(fold_auc)
        print(f"lgb seed={seed} fold={fold:02d} auc={fold_auc:.6f} best_iter={model.best_iteration_}")

best_p = 1.0
best_p_auc = -1.0
oof_candidates = [seed_oof[s] for s in LGB_SEEDS]
test_candidates = [seed_test[s] for s in LGB_SEEDS]
for p in [0.5, 0.75, 1.0, 1.25, 1.5, 2.0]:
    auc = roc_auc_score(y, power_mean(oof_candidates, p))
    if auc > best_p_auc:
        best_p_auc = auc
        best_p = p

oof_lgb = power_mean(oof_candidates, best_p)
test_lgb = power_mean(test_candidates, best_p)
print(f"lgb power-mean p={best_p:.2f} oof auc: {roc_auc_score(y, oof_lgb):.6f}")


def _fit_catboost_full(X_tr, y_tr, X_va, y_va, X_te, params):
    X_tr, X_va, X_te = add_group_income_residuals(
        X_tr, X_va, X_te, RESIDUAL_GROUP_COLS
    )
    X_tr, X_va, X_te = add_cell_income_isotonic(X_tr, y_tr, [X_va, X_te])
    X_tr, X_va, X_te = add_nested_income_te(
        X_tr, y_tr, [X_va, X_te], INCOME_TE_COLS, prior=200.0, suffix="200"
    )
    for prior in INCOME_TE_PRIORS:
        if abs(prior - 200.0) < 1e-9:
            continue
        X_tr, X_va, X_te = add_nested_income_te(
            X_tr,
            y_tr,
            [X_va, X_te],
            INCOME_TE_MULTI_PRIOR,
            prior=prior,
            suffix=str(int(prior)),
        )
    for col in CATBOOST_CAT_COLS:
        if col in X_tr.columns:
            X_tr[col] = X_tr[col].astype(str)
            X_va[col] = X_va[col].astype(str)
            X_te[col] = X_te[col].astype(str)
    cat_idx = get_cat_idx(X_tr.columns, CATBOOST_CAT_COLS)
    train_pool = Pool(X_tr, y_tr, cat_features=cat_idx)
    valid_pool = Pool(X_va, y_va, cat_features=cat_idx)
    test_pool = Pool(X_te, cat_features=cat_idx)
    try:
        model = CatBoostClassifier(**params)
        model.fit(train_pool, eval_set=valid_pool, early_stopping_rounds=120, use_best_model=True)
    except Exception as e:
        print(f"cb retry max_ctr_complexity=2 ({type(e).__name__})")
        model = CatBoostClassifier(**{**params, "max_ctr_complexity": 2})
        model.fit(train_pool, eval_set=valid_pool, early_stopping_rounds=120, use_best_model=True)
    return (
        model.predict_proba(valid_pool)[:, 1],
        model.predict_proba(test_pool)[:, 1],
        model.get_best_iteration(),
    )


def _fit_catboost_lean(X_tr, y_tr, X_va, y_va, X_te, params):
    """No nested TE / multi-prior — native cats + cell-iso only."""
    X_tr, X_va, X_te = add_group_income_residuals(
        X_tr, X_va, X_te, RESIDUAL_GROUP_COLS
    )
    X_tr, X_va, X_te = add_cell_income_isotonic(X_tr, y_tr, [X_va, X_te])
    # drop high-card TE string keys except lean cats; keep numeric engineered cols
    drop_keys = [c for c in INCOME_TE_COLS if c not in LEAN_CB_CATS and c in X_tr.columns]
    drop_keys += [c for c in ENCODE_COLS if c not in LEAN_CB_CATS and c in X_tr.columns]
    X_tr = X_tr.drop(columns=drop_keys, errors="ignore")
    X_va = X_va.drop(columns=drop_keys, errors="ignore")
    X_te = X_te.drop(columns=drop_keys, errors="ignore")
    for col in LEAN_CB_CATS:
        if col in X_tr.columns:
            X_tr[col] = X_tr[col].astype(str)
            X_va[col] = X_va[col].astype(str)
            X_te[col] = X_te[col].astype(str)
    cat_idx = get_cat_idx(X_tr.columns, LEAN_CB_CATS)
    train_pool = Pool(X_tr, y_tr, cat_features=cat_idx)
    valid_pool = Pool(X_va, y_va, cat_features=cat_idx)
    test_pool = Pool(X_te, cat_features=cat_idx)
    model = CatBoostClassifier(**params)
    model.fit(train_pool, eval_set=valid_pool, early_stopping_rounds=120, use_best_model=True)
    return (
        model.predict_proba(valid_pool)[:, 1],
        model.predict_proba(test_pool)[:, 1],
        model.get_best_iteration(),
    )


print(f"training {len(CB_SEEDS) * CB_SPLITS} catboost models")
for seed in CB_SEEDS:
    skf = StratifiedKFold(n_splits=CB_SPLITS, shuffle=True, random_state=seed)
    for fold, (train_idx, valid_idx) in enumerate(skf.split(X, y), start=1):
        X_tr = X.iloc[train_idx].reset_index(drop=True)
        X_va = X.iloc[valid_idx].reset_index(drop=True)
        y_tr, y_va = y[train_idx], y[valid_idx]
        params = {**CB_PARAMS, "random_seed": seed}
        valid_preds, te_preds, best_iter = _fit_catboost_full(
            X_tr, y_tr, X_va, y_va, X_test.copy(), params
        )
        fold_auc = roc_auc_score(y_va, valid_preds)
        cb_seed_oof[seed][valid_idx] = valid_preds
        cb_seed_test[seed] += te_preds / CB_SPLITS
        cb_scores.append(fold_auc)
        print(f"cb seed={seed} fold={fold:02d} auc={fold_auc:.6f} best_iter={best_iter}")

oof_cb = np.mean([cb_seed_oof[s] for s in CB_SEEDS], axis=0)
test_cb = np.mean([cb_seed_test[s] for s in CB_SEEDS], axis=0)
print(f"cb bag oof auc: {roc_auc_score(y, oof_cb):.6f}")

print(f"training {CB_SPLITS} lean catboost models (no nested TE)")
skf = StratifiedKFold(n_splits=CB_SPLITS, shuffle=True, random_state=LEAN_SEED)
for fold, (train_idx, valid_idx) in enumerate(skf.split(X, y), start=1):
    X_tr = X.iloc[train_idx].reset_index(drop=True)
    X_va = X.iloc[valid_idx].reset_index(drop=True)
    y_tr, y_va = y[train_idx], y[valid_idx]
    valid_preds, te_preds, best_iter = _fit_catboost_lean(
        X_tr, y_tr, X_va, y_va, X_test.copy(), {**LEAN_CB_PARAMS, "random_seed": LEAN_SEED + fold}
    )
    fold_auc = roc_auc_score(y_va, valid_preds)
    oof_lean[valid_idx] = valid_preds
    test_lean += te_preds / CB_SPLITS
    lean_scores.append(fold_auc)
    print(f"lean fold={fold:02d} auc={fold_auc:.6f} best_iter={best_iter}")

print(f"lean cb oof auc: {roc_auc_score(y, oof_lean):.6f}")

print(f"training {len(XGB_SEEDS) * XGB_SPLITS} xgboost models")
for seed in XGB_SEEDS:
    skf = StratifiedKFold(n_splits=XGB_SPLITS, shuffle=True, random_state=seed)
    for fold, (train_idx, valid_idx) in enumerate(skf.split(X, y), start=1):
        X_tr, X_va = X.iloc[train_idx], X.iloc[valid_idx]
        y_tr, y_va = y[train_idx], y[valid_idx]
        X_tr, X_va, X_te = prepare_te_frames(
            X_tr, y_tr, X_va, X_test, encode_cols=LEAN_ENCODE_COLS
        )
        X_tr, X_va, X_te = _drop_strings_for_trees(X_tr, X_va, X_te)
        model = xgb.XGBClassifier(
            **XGB_PARAMS, random_state=seed, n_jobs=-1, verbosity=0
        )
        model.fit(
            X_tr,
            y_tr,
            eval_set=[(X_va, y_va)],
            verbose=False,
        )
        vp = model.predict_proba(X_va)[:, 1]
        xgb_seed_oof[seed][valid_idx] = vp
        xgb_seed_test[seed] += model.predict_proba(X_te)[:, 1] / XGB_SPLITS
        fold_auc = roc_auc_score(y_va, vp)
        xgb_scores.append(fold_auc)
        print(
            f"xgb seed={seed} fold={fold:02d} auc={fold_auc:.6f} "
            f"best_iter={model.best_iteration}"
        )

oof_xgb = np.mean([xgb_seed_oof[s] for s in XGB_SEEDS], axis=0)
test_xgb = np.mean([xgb_seed_test[s] for s in XGB_SEEDS], axis=0)
print(f"xgb oof auc: {roc_auc_score(y, oof_xgb):.6f}")



def to_logit(p):
    p = np.clip(p, 1e-6, 1 - 1e-6)
    return np.log(p / (1.0 - p))


def rank_avg(arrays):
    ranks = [rankdata(a) / (len(a) + 1.0) for a in arrays]
    return np.mean(ranks, axis=0)


def cellwise_calibrate(oof_p, test_p, y_true, key_tr, key_te, n_splits=5, seed=99):
    """Nested isotonic: map model scores -> y inside each subsidy|env cell."""
    from sklearn.isotonic import IsotonicRegression

    oof_c = np.zeros(len(y_true), dtype=np.float64)
    skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=seed)
    for tr, va in skf.split(np.zeros(len(y_true)), y_true):
        pred = oof_p[va].copy()
        for cell in np.unique(key_tr[tr]):
            mtr = key_tr[tr] == cell
            mva = key_tr[va] == cell
            if mtr.sum() < 80 or len(np.unique(y_true[tr][mtr])) < 2 or not mva.any():
                continue
            iso = IsotonicRegression(out_of_bounds="clip")
            iso.fit(oof_p[tr][mtr], y_true[tr][mtr])
            pred[mva] = iso.predict(oof_p[va][mva])
        oof_c[va] = pred

    # full-data maps for test
    test_c = test_p.copy()
    for cell in np.unique(key_tr):
        mtr = key_tr == cell
        mte = key_te == cell
        if mtr.sum() < 80 or len(np.unique(y_true[mtr])) < 2 or not mte.any():
            continue
        iso = IsotonicRegression(out_of_bounds="clip")
        iso.fit(oof_p[mtr], y_true[mtr])
        test_c[mte] = iso.predict(test_p[mte])
    return oof_c, test_c


# Main linear blend of LGB + full CB + lean CB + XGB
best_lin_auc = -1.0
best_lin_w = (0.35, 0.25, 0.2, 0.2)
for w0 in np.linspace(0.15, 0.60, 10):
    for w1 in np.linspace(0.10, 0.45, 8):
        for w2 in np.linspace(0.05, 0.35, 7):
            w3 = 1.0 - w0 - w1 - w2
            if w3 < 0.05 or w3 > 0.40:
                continue
            blend = w0 * oof_lgb + w1 * oof_cb + w2 * oof_lean + w3 * oof_xgb
            auc = roc_auc_score(y, blend)
            if auc > best_lin_auc:
                best_lin_auc = auc
                best_lin_w = (w0, w1, w2, w3)

w_lgb, w_cb, w_lean, w_xgb = best_lin_w
oof_linear = w_lgb * oof_lgb + w_cb * oof_cb + w_lean * oof_lean + w_xgb * oof_xgb
test_linear = w_lgb * test_lgb + w_cb * test_cb + w_lean * test_lean + w_xgb * test_xgb

oof_cal, test_cal = cellwise_calibrate(
    oof_linear, test_linear, y, CELL_KEY_TRAIN, CELL_KEY_TEST
)
cal_auc = roc_auc_score(y, oof_cal)
print(f"cell-calibrated linear auc: {cal_auc:.6f} (delta {cal_auc - best_lin_auc:+.6f})")

stack_oof = np.column_stack(
    [to_logit(seed_oof[s]) for s in LGB_SEEDS]
    + [to_logit(cb_seed_oof[s]) for s in CB_SEEDS]
    + [to_logit(oof_lean)]
    + [to_logit(xgb_seed_oof[s]) for s in XGB_SEEDS]
)
stack_test = np.column_stack(
    [to_logit(seed_test[s]) for s in LGB_SEEDS]
    + [to_logit(cb_seed_test[s]) for s in CB_SEEDS]
    + [to_logit(test_lean)]
    + [to_logit(xgb_seed_test[s]) for s in XGB_SEEDS]
)
meta = LogisticRegression(C=0.5, max_iter=3000, solver="lbfgs")
meta.fit(stack_oof, y)
oof_ridge = meta.predict_proba(stack_oof)[:, 1]
test_ridge = meta.predict_proba(stack_test)[:, 1]
ridge_auc = roc_auc_score(y, oof_ridge)

# also calibrate meta
oof_ridge_cal, test_ridge_cal = cellwise_calibrate(
    oof_ridge, test_ridge, y, CELL_KEY_TRAIN, CELL_KEY_TEST
)
ridge_cal_auc = roc_auc_score(y, oof_ridge_cal)

oof_rank = rank_avg([oof_lgb, oof_cb, oof_lean, oof_xgb])
test_rank = rank_avg([test_lgb, test_cb, test_lean, test_xgb])
rank_auc = roc_auc_score(y, oof_rank)

candidates = {
    "linear": (best_lin_auc, oof_linear, test_linear),
    "linear_cellcal": (cal_auc, oof_cal, test_cal),
    "logit_meta": (ridge_auc, oof_ridge, test_ridge),
    "logit_meta_cellcal": (ridge_cal_auc, oof_ridge_cal, test_ridge_cal),
    "rank_avg": (rank_auc, oof_rank, test_rank),
}
best_name = max(candidates, key=lambda k: candidates[k][0])
oof_auc, oof_preds, test_preds = candidates[best_name]

print(f"lgb mean fold: {np.mean(lgb_scores):.6f}")
print(f"cb mean fold: {np.mean(cb_scores):.6f}")
print(f"lean mean fold: {np.mean(lean_scores):.6f}")
print(f"xgb mean fold: {np.mean(xgb_scores):.6f}")
print(
    f"linear weights lgb/cb/lean/xgb: {w_lgb:.3f}/{w_cb:.3f}/{w_lean:.3f}/{w_xgb:.3f} "
    f"auc={best_lin_auc:.6f}"
)
print(f"logit meta auc: {ridge_auc:.6f} coef={np.round(meta.coef_[0], 4)}")
print(f"logit meta cellcal auc: {ridge_cal_auc:.6f}")
print(f"rank avg auc: {rank_auc:.6f}")
print(f"selected blend: {best_name}")
print(f"final oof auc: {oof_auc:.6f}")


PUB_DIR = Path("external/pub_lgb94607")
MEGA_DIR = Path("external/mega_lgb94633")
NAJI_DIR = Path("external/naji_lgb94638")
HERM_DIR = Path("external/herm_xgb94627")
FV_DIR = Path("external/mega_four_views")

oof_pub = np.load(PUB_DIR / "oof_predictions_lgb_42_5.npy")
test_pub = np.load(PUB_DIR / "test_predictions_lgb_42_5.npy")
oof_mega = np.load(MEGA_DIR / "oof_hybrid.npy")
test_mega = np.load(MEGA_DIR / "test_hybrid.npy")

oof_naji = (
    pd.read_csv(NAJI_DIR / "oof.csv").set_index("id").reindex(train["id"])["OOF_Pred"].to_numpy(
        dtype=np.float64
    )
)
test_naji = (
    pd.read_csv(NAJI_DIR / "test.csv")
    .set_index("id")
    .reindex(test["id"])["Will_Buy_EV"]
    .to_numpy(dtype=np.float64)
)
oof_herm = np.load(HERM_DIR / "oof.npy").astype(np.float64).ravel()
test_herm = np.load(HERM_DIR / "test.npy").astype(np.float64).ravel()

fv_oof = pd.read_csv(FV_DIR / "oof.csv").set_index("id").reindex(train["id"])
fv_te = pd.read_csv(FV_DIR / "test.csv").set_index("id").reindex(test["id"])
oof_fvA = fv_oof["A_lgbm_triple_te_3seed"].to_numpy(dtype=np.float64)
test_fvA = fv_te["A_lgbm_triple_te_3seed"].to_numpy(dtype=np.float64)
oof_fvD = fv_oof["D_lgbm_alt_ladder_noexact"].to_numpy(dtype=np.float64)
test_fvD = fv_te["D_lgbm_alt_ladder_noexact"].to_numpy(dtype=np.float64)
oof_fvE = fv_oof["ensemble_ABCD"].to_numpy(dtype=np.float64)
test_fvE = fv_te["ensemble_ABCD"].to_numpy(dtype=np.float64)

ext_bank = {
    "pub": (oof_pub, test_pub),
    "mega": (oof_mega, test_mega),
    "naji": (oof_naji, test_naji),
    "herm": (oof_herm, test_herm),
    "fvA": (oof_fvA, test_fvA),
    "fvD": (oof_fvD, test_fvD),
    "fvE": (oof_fvE, test_fvE),
}
for name, (oo, _) in ext_bank.items():
    print(f"ext {name} oof auc: {roc_auc_score(y, oo):.6f}")

base_name = best_name
base_oof_auc = float(oof_auc)
base_oof = oof_preds.copy()
base_test = test_preds.copy()

ref = pd.read_csv("score94595.csv").set_index("id").reindex(test["id"])["Will_Buy_EV"].to_numpy(
    dtype=np.float64
)


def _rank01(a):
    return rankdata(a) / (len(a) + 1.0)


def _best_pair_lin(a, b, w_grid):
    best_w, best_auc = 0.5, -1.0
    for w in w_grid:
        auc = roc_auc_score(y, (1.0 - w) * a + w * b)
        if auc > best_auc:
            best_auc = auc
            best_w = float(w)
    return best_w, best_auc


def _best_pair_rank(a, b, w_grid):
    ra, rb = _rank01(a), _rank01(b)
    best_w, best_auc = 0.5, -1.0
    for w in w_grid:
        auc = roc_auc_score(y, (1.0 - w) * ra + w * rb)
        if auc > best_auc:
            best_auc = auc
            best_w = float(w)
    return best_w, best_auc


w_grid = np.linspace(0.15, 0.85, 29)
ext_cands = {}

# pair blends: base + each external
for name, (oo, te) in ext_bank.items():
    w, auc = _best_pair_lin(base_oof, oo, w_grid)
    ext_cands[f"lin_{name}_{w:.2f}"] = (
        auc,
        (1.0 - w) * base_oof + w * oo,
        (1.0 - w) * base_test + w * te,
    )
    wr, aucr = _best_pair_rank(base_oof, oo, w_grid)
    ext_cands[f"rank_{name}_{wr:.2f}"] = (
        aucr,
        (1.0 - wr) * _rank01(base_oof) + wr * _rank01(oo),
        (1.0 - wr) * _rank01(base_test) + wr * _rank01(te),
    )
    ext_cands[f"{name}_only"] = (roc_auc_score(y, oo), oo, te)

# focused 3-way: base + mega + herm (arch diversity) and base + fvE + herm
for tag, k1, k2 in [
    ("mega_herm", "mega", "herm"),
    ("fvE_herm", "fvE", "herm"),
    ("mega_fvD", "mega", "fvD"),
    ("naji_herm", "naji", "herm"),
]:
    o1, t1 = ext_bank[k1]
    o2, t2 = ext_bank[k2]
    best_auc, best_w = -1.0, (0.5, 0.25, 0.25)
    for w0 in np.linspace(0.25, 0.70, 10):
        for w1 in np.linspace(0.10, 0.50, 9):
            w2 = 1.0 - w0 - w1
            if w2 < 0.08 or w2 > 0.50:
                continue
            blend = w0 * base_oof + w1 * o1 + w2 * o2
            auc = roc_auc_score(y, blend)
            if auc > best_auc:
                best_auc = auc
                best_w = (w0, w1, w2)
    w0, w1, w2 = best_w
    ext_cands[f"lin3_{tag}_{w0:.2f}_{w1:.2f}_{w2:.2f}"] = (
        best_auc,
        w0 * base_oof + w1 * o1 + w2 * o2,
        w0 * base_test + w1 * t1 + w2 * t2,
    )
    # rank 3-way
    best_auc, best_w = -1.0, (0.5, 0.25, 0.25)
    r0, r1, r2 = _rank01(base_oof), _rank01(o1), _rank01(o2)
    rt0, rt1, rt2 = _rank01(base_test), _rank01(t1), _rank01(t2)
    for w0 in np.linspace(0.25, 0.70, 10):
        for w1 in np.linspace(0.10, 0.50, 9):
            w2 = 1.0 - w0 - w1
            if w2 < 0.08 or w2 > 0.50:
                continue
            blend = w0 * r0 + w1 * r1 + w2 * r2
            auc = roc_auc_score(y, blend)
            if auc > best_auc:
                best_auc = auc
                best_w = (w0, w1, w2)
    w0, w1, w2 = best_w
    ext_cands[f"rank3_{tag}_{w0:.2f}_{w1:.2f}_{w2:.2f}"] = (
        best_auc,
        w0 * r0 + w1 * r1 + w2 * r2,
        w0 * rt0 + w1 * rt1 + w2 * rt2,
    )

# logit meta over base + diverse externals
meta_keys = ["mega", "fvE", "herm", "fvD", "naji", "pub"]
stack_ext_oof = np.column_stack(
    [to_logit(base_oof)] + [to_logit(ext_bank[k][0]) for k in meta_keys]
)
stack_ext_te = np.column_stack(
    [to_logit(base_test)] + [to_logit(ext_bank[k][1]) for k in meta_keys]
)
meta_ext = LogisticRegression(C=0.35, max_iter=4000, solver="lbfgs")
meta_ext.fit(stack_ext_oof, y)
oof_meta_ext = meta_ext.predict_proba(stack_ext_oof)[:, 1]
test_meta_ext = meta_ext.predict_proba(stack_ext_te)[:, 1]
meta_ext_auc = roc_auc_score(y, oof_meta_ext)
ext_cands["logit_meta_ext"] = (meta_ext_auc, oof_meta_ext, test_meta_ext)
print(
    f"logit_meta_ext auc: {meta_ext_auc:.6f} "
    f"coef={np.round(meta_ext.coef_[0], 4)}"
)

best_ext = max(ext_cands, key=lambda k: ext_cands[k][0])
oof_auc, oof_preds, test_preds = ext_cands[best_ext]
best_name = f"{base_name}+{best_ext}"

RTAB_DIR = Path("external/realmlp_tab94620")
RMLP_DIR = Path("external/realmlp_pt94621")
NCB_DIR = Path("external/naji_cb94624")
oof_rtab = (
    pd.read_csv(RTAB_DIR / "oof.csv")
    .set_index("id")
    .reindex(train["id"])["Will_Buy_EV"]
    .to_numpy(dtype=np.float64)
)
test_rtab = (
    pd.read_csv(RTAB_DIR / "test.csv")
    .set_index("id")
    .reindex(test["id"])["Will_Buy_EV"]
    .to_numpy(dtype=np.float64)
)
oof_rmlp = (
    pd.read_csv(RMLP_DIR / "oof.csv")
    .set_index("id")
    .reindex(train["id"])["Will_Buy_EV"]
    .to_numpy(dtype=np.float64)
)
test_rmlp = (
    pd.read_csv(RMLP_DIR / "test.csv")
    .set_index("id")
    .reindex(test["id"])["Will_Buy_EV"]
    .to_numpy(dtype=np.float64)
)
oof_ncb = (
    pd.read_csv(NCB_DIR / "oof.csv")
    .set_index("id")
    .reindex(train["id"])["OOF_Pred"]
    .to_numpy(dtype=np.float64)
)
test_ncb = (
    pd.read_csv(NCB_DIR / "test.csv")
    .set_index("id")
    .reindex(test["id"])["Will_Buy_EV"]
    .to_numpy(dtype=np.float64)
)

resid_bank = {
    "rtab": (oof_rtab, test_rtab),
    "rmlp": (oof_rmlp, test_rmlp),
    "ncb": (oof_ncb, test_ncb),
}
r_sel = rankdata(oof_preds) / (len(oof_preds) + 1.0)
rt_sel = rankdata(test_preds) / (len(test_preds) + 1.0)
best_res = (oof_auc, None, 0.0, oof_preds, test_preds)
for rname, (ro, rt) in resid_bank.items():
    rr = rankdata(ro) / (len(ro) + 1.0)
    rrt = rankdata(rt) / (len(rt) + 1.0)
    for w in np.linspace(0.04, 0.15, 12):
        blend = (1.0 - w) * r_sel + w * rr
        auc = roc_auc_score(y, blend)
        if auc > best_res[0] + 1e-7:
            best_res = (
                auc,
                rname,
                float(w),
                blend,
                (1.0 - w) * rt_sel + w * rrt,
            )
if best_res[1] is not None:
    oof_auc, rname, rw, oof_preds, test_preds = best_res
    best_name = f"{best_name}+resid_{rname}_{rw:.2f}"
    print(f"constrained residual: {rname} w={rw:.3f} auc={oof_auc:.6f}")
else:
    print("constrained residual: none (no OOF gain within w<=0.15)")

sp = float(spearmanr(test_preds, ref).correlation)
print(
    f"ext selected={best_name} oof={oof_auc:.6f} "
    f"(base {base_oof_auc:.6f}, delta {oof_auc - base_oof_auc:+.6f}) "
    f"spearman_vs_94595={sp:.6f}"
)
print(f"final oof auc: {oof_auc:.6f}")
```
